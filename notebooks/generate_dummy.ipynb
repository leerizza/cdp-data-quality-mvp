```python
from faker import Faker
import pandas as pd
import random

fake = Faker("id_ID")

rows = []

for i in range(1000):

    nik = str(random.randint(3200000000000000,3299999999999999))

    rows.append({

        "customer_id": f"C{i:05d}",

        "nik": nik,

        "full_name": fake.name(),

        "phone": "08"+str(random.randint(1111111111,9999999999)),

        "email": fake.email(),

        "birth_date": fake.date_of_birth(minimum_age=18,maximum_age=60)

    })

df = pd.DataFrame(rows)

# Inject bad data

df.loc[10,"nik"] = None

df.loc[20,"phone"] = "ABC"

df.loc[30,"birth_date"] = "2035-01-01"

df.loc[40,"nik"] = df.loc[0,"nik"]

df.to_csv("data/customer.csv",index=False)

print(df.head())
```
